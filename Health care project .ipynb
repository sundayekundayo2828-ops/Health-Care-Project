```python
PROBLEM 3: Patient Demographics & Age Segmentation
```

```python
import pandas as pd

# 1. Define the exact path to your dataset
file_path = r"C:\Users\sunda\Desktop\Health Care Project\Data\patient_records.csv"

# 2. Load the dataset
df = pd.read_csv(file_path)

# 3. Convert Date of Birth string values to datetime objects
df['Date_of_Birth'] = pd.to_datetime(df['Date_of_Birth'])

# 4. Calculate current age based on the fixed project baseline year (2026)
current_year = 2026
df['Age'] = current_year - df['Date_of_Birth'].dt.year

# 5. Define clinical age brackets and segment the patients
age_bins = [0, 17, 34, 49, 64, 120]
age_labels = ['0-17 (Pediatric)', '18-34 (Young Adult)', '35-49 (Adult)', '50-64 (Middle Aged)', '65+ (Senior)']
df['Age_Group'] = pd.cut(df['Age'], bins=age_bins, labels=age_labels)

# 6. Generate the demographic summary table
demographic_summary = df['Age_Group'].value_counts().sort_index()

# 7. Display the final output
print("--- Patient Age Demographics Summary ---")
print(demographic_summary)
```

```python
4. Medication Inventory & Demand Forecasting
```

```python
import pandas as pd

# 1. Define the exact path to your dataset
file_path = r"C:\Users\sunda\Desktop\Health Care Project\Data\patient_records.csv"

# 2. Load the dataset
df = pd.read_csv(file_path)

# 3. Filter out patients who are not currently on any medication
active_prescriptions = df[df['Prescribed_Medication'] != 'None']

# 4. Cross-tabulate Medication volume against the conditions they treat
medication_forecast = pd.crosstab(
    index=active_prescriptions['Prescribed_Medication'],
    columns=active_prescriptions['Diagnosed_Condition'],
    margins=True, 
    margins_name="Total_Demand"
)

# 5. Display the final output
print("--- Medication Demand Matrix ---")
print(medication_forecast)
```

```python
PROBLEM 5: Patient Retention & Appointment Recency Tracking
```

```python
import pandas as pd

# 1. Define the exact path to your dataset
file_path = r"C:\Users\sunda\Desktop\Health Care Project\Data\patient_records.csv"

# 2. Load the dataset
df = pd.read_csv(file_path)

# 3. Convert Last Visit Date string values to datetime objects
df['Last_Visit_Date'] = pd.to_datetime(df['Last_Visit_Date'])

# 4. Find the most recent date in the entire dataset to establish a baseline
baseline_date = df['Last_Visit_Date'].max()

# 5. Calculate how many days have passed since each patient's last visit
df['Days_Since_Last_Visit'] = (baseline_date - df['Last_Visit_Date']).dt.days

# 6. Categorize appointment status based on recency
def check_retention(days):
    if days <= 90:
        return 'Active (Within 3 months)'
    elif days <= 180:
        return 'Standard (3-6 months)'
    else:
        return 'Lapsed (Over 6 months)'

df['Retention_Status'] = df['Days_Since_Last_Visit'].apply(check_retention)

# 7. Generate operational retention report
retention_report = df['Retention_Status'].value_counts()

# 8. Display the final output
print(f"--- Patient Retention Status Report (Baseline Date: {baseline_date.strftime('%Y-%m-%d')}) ---")
print(retention_report)

```
